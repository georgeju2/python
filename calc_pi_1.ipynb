```python
N= 100000
M = 10000000
arr = np.random.randint(M, size=(N, 2))
```

```python
count = 0
for i in range(N):
    #print(arr[i])
    gcd =  math.gcd(*arr[i])
    if gcd==1: count +=1
    
math.sqrt(6/ (count / N))
```
