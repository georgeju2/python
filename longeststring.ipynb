```python
import random
count_d = {}
def init(d_size):
    for i in range(d_size+1):
        count_d[i] = 0
    
```

```python
def get_loop(n):
    #print("---------------")
    lst = random.sample(range(n), k=n)
    left_set = set(range(n))
    start_pos = 0
    idx = start_pos
    looplst = []
    newlst, count = [], 0
    while count < len(lst):
        idx = lst[idx]
        newlst.append(idx)
        count += 1
        if idx==start_pos: 
            looplst.append(newlst)
            #print("find list: ", len(newlst), newlst)
            left_set = left_set.difference(newlst)
            if len(left_set)==0: break
            #print("left", len(left_set), left_set)
            newlst = []
            min_left = min(left_set)
            idx, start_pos = min_left, min_left
            #print(*looplst, sep='\n')
    loopsz = [len(x) for x in looplst]
    #print(loopsz)
    maxloopsz = max(loopsz)
    if maxloopsz == 98 and len(looplst[0])!=98:
        print(len(looplst), looplst)
    count_d[maxloopsz] = count_d[maxloopsz]+1
    # print(lst)
    # print(newlst)

init(100)
N = 100000
for i in range(N):
    get_loop(100)
for i in range(50, 100):
    print(i, count_d[i]/N*100)
```
