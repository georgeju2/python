```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        last_seen = {}
        max_len, run_len= 0, 0
        for i, c in enumerate(s):
            print(i, c, end= ' ')
            run_len += 1
            if c in last_seen:
                upb = i - last_seen[c] 
                print ("upb = ", upb)
                run_len = min(upb, run_len)
            last_seen[c] = i
            if run_len > max_len : max_len = run_len
            print(run_len, max_len)
        return max_len
```

```python
sol = Solution()
```

```python
s = "abcabcbb"
sol.lengthOfLongestSubstring(s)
```
